```python
from rdflib import Graph, Namespace, URIRef, RDF, OWL, RDFS
from owlrl import DeductiveClosure, OWLRL_Semantics
```

```python
# Load existing TTL file
g = Graph()
g.parse("example_rdf.ttl", format="turtle")
```

```python
ex = Namespace("http://example.org/")
g.bind("ex", ex)
g.bind("owl", OWL)
g.bind("rdfs", RDFS)
```

```python
# Disjoint: Person vs Organization
g.add((ex.Person, RDF.type, OWL.Class))
g.add((ex.Organization, RDF.type, OWL.Class))
g.add((ex.Person, OWL.disjointWith, ex.Organization))

# Property types
g.add((ex.wasLyingTo, RDF.type, OWL.IrreflexiveProperty))
g.add((ex.hasBusinessPartner, RDF.type, OWL.SymmetricProperty))
g.add((ex.adviserTo, RDF.type, OWL.AsymmetricProperty))
g.add((ex.presidentOf, RDF.type, OWL.FunctionalProperty))
g.add((ex.hasPresident, RDF.type, OWL.InverseFunctionalProperty))
g.add((ex.supporting, RDF.type, OWL.TransitiveProperty))

# Subproperties
g.add((ex.adviserTo, RDFS.subPropertyOf, ex.supporting))
g.add((ex.campaignManager, RDFS.subPropertyOf, ex.supporting))

# Subclass relationships
g.add((ex.TaxEvasion, RDFS.subClassOf, ex.BankAndTaxFraud))
g.add((ex.RepublicanPolitician, RDFS.subClassOf, ex.Politician))
g.add((ex.RepublicanPolitician, RDFS.subClassOf, ex.Republican))

# Domain and range to connect the ontology
g.add((ex.wasLyingTo, RDFS.domain, ex.Person))
g.add((ex.wasLyingTo, RDFS.range, ex.Organization))
g.add((ex.adviserTo, RDFS.domain, ex.Person))
g.add((ex.adviserTo, RDFS.range, ex.Person))
```

```python
# Run OWL RL reasoning
DeductiveClosure(OWLRL_Semantics).expand(g)

# Serialize the updated graph to Turtle
g.serialize(destination="extended_with_owl.xml", format="xml")

print("Reasoning complete. Output saved to extended_with_owl.xml")
```
