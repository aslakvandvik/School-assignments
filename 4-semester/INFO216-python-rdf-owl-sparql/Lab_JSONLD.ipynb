```python
from rdflib import Graph

# Create an RDF graph
g = Graph()

# Parse JSON-LD data from ConceptNet
url = "https://api.conceptnet.io/c/en/indictment"
g.parse(url, format="json-ld")

# Count the number of triples
print(f"Number of triples: {len(g)}")

# Print all triples
for s, p, o in g:
    print(s, p, o)
```

```python
from rdflib.plugins.sparql import prepareQuery

# Query distinct predicates
query_predicates = prepareQuery("""
    SELECT DISTINCT ?p WHERE {
        ?s ?p ?o .
    }
""")

print("Distinct Predicates:")
for row in g.query(query_predicates):
    print(row.p)

# Query distinct types
query_types = prepareQuery("""
    SELECT DISTINCT ?o WHERE {
        ?s rdf:type ?o .
    }
""")

print("\nDistinct Types:")
for row in g.query(query_types):
    print(row.o)
```

```python
import json
import requests

# Download JSON-LD data from ConceptNet
CN_BASE = "http://api.conceptnet.io/c/en/"
json_obj = requests.get(CN_BASE + "indictment").json()

# Define a custom context for simplification
json_obj["@context"] = {
    "edges": "http://ex.org/t",
    "start": "http://ex.org/s",
    "rel": "http://ex.org/p",
    "end": "http://ex.org/o",
    "label": "http://www.w3.org/2000/01/rdf-schema#label",
    "@base": "http://ex.org/"
}

# Convert modified JSON object to a string
json_str = json.dumps(json_obj)

# Save JSON to avoid repeated API calls
with open("conceptnet_indictment.json", "w") as f:
    f.write(json_str)
```

```python
# Create a new graph
g2 = Graph()

# Parse from the JSON string
g2.parse(data=json_str, format="json-ld")

print(f"New Graph: Number of triples: {len(g2)}")

# Print the triples again
for s, p, o in g2:
    print(s, p, o)
```

```python
query_all_triples = prepareQuery("""
    SELECT ?s ?p ?o WHERE {
        ?s ?p ?o .
    }
""")

print("\nAll Triples:")
for row in g2.query(query_all_triples):
    print(row.s, row.p, row.o)
```

```python
query_with_labels = prepareQuery("""
    SELECT ?s ?sLabel ?p ?o ?oLabel WHERE {
        ?s ?p ?o .
        OPTIONAL { ?s <http://www.w3.org/2000/01/rdf-schema#label> ?sLabel . }
        OPTIONAL { ?o <http://www.w3.org/2000/01/rdf-schema#label> ?oLabel . }
    }
""")

print("\nTriples with Labels:")
for row in g2.query(query_with_labels):
    print(row.s, row.sLabel, row.p, row.o, row.oLabel)
```

```python
query_construct = prepareQuery("""
    CONSTRUCT {
        ?s ?p ?o .
    } WHERE {
        ?s ?p ?o .
    }
""")

# Create a new graph to store the results
g3 = Graph()
g3 += g2.query(query_construct)

# Save to a file
g3.serialize("filtered_conceptnet.ttl", format="turtle")

print("\nConstructed graph saved as 'filtered_conceptnet.ttl'.")
```
