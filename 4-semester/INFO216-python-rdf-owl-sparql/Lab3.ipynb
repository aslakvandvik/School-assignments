```python
import rdflib
```

```python
# Load the graph
g = rdflib.Graph()
g.parse("russia_investigation_kg.ttl", format="ttl")
```

```python
print("Task: Print out a list of all the predicates used in your graph.")

# Get all predicates in the graph
predicates = set()
for _, predicate, _ in g:
    predicates.add(predicate)

# Print the sorted list of predicates
for predicate in sorted(predicates):
    print(predicate)
```

```python
print("Task: Print out a sorted list of all the presidents represented in your graph.")

# Find all the presidents in the graph
presidents = set()
for subject, predicate, obj in g:
    if "president" in str(predicate):  # Look for the president property
        presidents.add(str(obj))  # The object of this predicate is the president

# Print sorted list of presidents
for president in sorted(presidents):
    print(president)
```
