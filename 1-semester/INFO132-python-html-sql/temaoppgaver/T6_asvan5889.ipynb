```python
# Oppgave 1

def lag_fortegnelse(personer):
    fortegnelse = {}
    for person in personer:
        navn, alder = person
        fortegnelse[navn] = alder
    return fortegnelse

lag_fortegnelse([('Alice', 25), ('Bob', 30), ('Charlie', 35)])
```

```python
# Oppgave 2
def sorter_personer_etter_navn(personer):
    return sorted(personer.items())

sorter_personer_etter_navn({'Alice': 25, 'Charlie': 35, 'Bob': 30})
```

```python
# Oppgave 3
def finn_venners_venner(venneliste, navn):
    gjensidige_venner = []

    for person, venneliste_i in venneliste.items():
        if navn in venneliste_i and person != navn:
            gjensidige_venner.append(person)

    return gjensidige_venner


vennefortegnelse = {
                    'Alice': ['Bob', 'Charlie', 'Devin'], 
                    'Bob': ['Charlie'],
                    'Charlie': ['Alice'], 
                    'Devin': ['Alice', 'Bob']
                    }

print(finn_venners_venner(vennefortegnelse, 'Alice'))
print(finn_venners_venner(vennefortegnelse, 'Bob'))
print(finn_venners_venner(vennefortegnelse, 'Charlie'))
print(finn_venners_venner(vennefortegnelse, 'Devin'))
```
