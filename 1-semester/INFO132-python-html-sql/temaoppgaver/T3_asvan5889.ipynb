```python
# Oppgave 1a)

def multiplikasjonstabell(n):
    
    # Beregner bredden ved å finne antall sifre i tabellen, der største tall er n*n
    max_bredde = len(str(n*n))

    # Skrive ut den enkle tabellen a)
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            print(f"{i*j:>{max_bredde}}", end = " ")
        print()

#Test
multiplikasjonstabell(3)
print("")
multiplikasjonstabell(5)
print("")

# Oppgave 1b)

def multiplikasjonstabell_fin(n):
    print(" * |", end=" ")
    for i in range(1, n + 1):
        print(f"{i:2}", end=" ")
    print("\n---+" + "-" * (n * 3))

    for i in range(1, n + 1):
        print(f"{i:2} |", end= " ")
        for j in range(1, n + 1):
            print(f"{i * j:2}", end=" ")
        print()

#Test
multiplikasjonstabell_fin(3)
print("")
multiplikasjonstabell_fin(10)
```

```python
# Oppgave 2a)

def for_fakultet(n):
    if n < 0:
        return f"Fakultetet av {n} er ikke definert"
    elif n == 0:
        return 1
    else:
        resultat = 1
        for i in range(1, n + 1):
            resultat *= i
        return resultat
    
#Test 
print(for_fakultet(1))
print(for_fakultet(2))
print(for_fakultet(3))
print(for_fakultet(4))
print(for_fakultet(0))
print(for_fakultet(-3.14159)) 
```

```python
# Oppgave 2b) 

def while_fakultet(n):
    if n < 0:
        return f"Fakultetet av {n} er ikke definert"
    elif n == 0:
        return 1
    else:
        resultat = 1
        i = 1
        while i <= n:
            resultat *= i
            i += 1
        return resultat

#Test
print(while_fakultet(1))
print(while_fakultet(2))
print(while_fakultet(3))
print(while_fakultet(4))
print(while_fakultet(0))
print(while_fakultet(-3.14159))
```
