# Oppgave 3

```python
# Oppgave 3. 
class Bil:
    def __init__(self, merke, modell, ar):
        self.merke = merke
        self.modell = modell
        self.ar = ar

    def skriv_ut_info(self):
        return f'Merke: {self.merke}, Modell: {self.modell}, År: {self.ar}'

    def alder(self, nåværende_ar=2023):
        return nåværende_ar - self.ar

bil1 = Bil('Volkswagen', 'Golf', 2019)
bil2 = Bil('Tesla', 'Model 3', 2021)

print(bil1.alder())
print(bil2.alder(2021))
```

# Oppgave 4

```python
# Oppgave 4. 
def bilpark_info(bilpark):
    for bil in bilpark:
        print(bil.skriv_ut_info())

bilpark = [bil1, bil2]
bilpark_info(bilpark)
```

# Oppgave 5

```python
# Oppgave 5. 
class Eier:
    def __init__(self, navn, adresse):
        self.navn = navn
        self.adresse = adresse

    def skriv_ut_info(self):
        return f'Navn: {self.navn}, Adresse: {self.adresse}'

eier = Eier('Ola Nordmann', 'Osloveien 1')
print(eier.skriv_ut_info())
```

# Oppgave 6

```python
# Oppgave 6. 
class Bil:
    def __init__(self, merke, modell, år):
        self.merke = merke
        self.modell = modell
        self.år = år
        self.eier = None  # Setter eier til standardverdi None

    def skriv_ut_info(self):
        info = f'Merke: {self.merke}, Modell: {self.modell}, År: {self.år}'
        if self.eier:
            info += f', Eier: {self.eier.skriv_ut_info()}'
        return info

bil1 = Bil('Volkswagen', 'Golf', 2019)
bil2 = Bil('Tesla', 'Model 3', 2021)

eier = Eier('Ola Nordmann', 'Osloveien 1')
bil1.eier = eier

print(bil1.skriv_ut_info())
print('')

bilpark = [bil1, bil2]
bilpark_info(bilpark)
```
