```python
# Oppgave 1. 

def telle_bokstaver(string, bokstav): #a) 
    return print(string.count(bokstav))

def telle_ord(string): #b)
    telte_ord = string.split()
    return print(len(telte_ord))


telle_bokstaver('pannekaker','n')
telle_ord('Jeg heter Aslak')
```

```python
# Oppgave 2. 

def erstatte_bokstav(string, erstatt_bokstav, ny_bokstav):
    return print(string.replace(erstatt_bokstav, ny_bokstav))

erstatte_bokstav('Hunden min heter Emma', 'm', 'r')
```

```python
# Oppgave 3. 

def formater_navn(fornavn, etternavn, mellomnavn = None):
    if mellomnavn: 
        return print(f'{etternavn}, {fornavn} {mellomnavn}')
    else:
        return print(f'{etternavn}, {fornavn}')
    
formater_navn('Aslak', 'Vandvik')
print("")
formater_navn('Aslak', 'Vandvik', 'Prestvik')

```

# Oppgave 4. 

```python
# Oppgave 4. 
def finn_mest_brukte_ord(setning):
    ord_liste = setning.split()
    mest_brukte_ord = None
    maks_telling = 0
    
    for ord in ord_liste:
        opptelling = ord_liste.count(ord)
        if opptelling > maks_telling:
            mest_brukte_ord = ord
            maks_telling = opptelling
    
    return print(mest_brukte_ord)

finn_mest_brukte_ord('Jeg liker å gå på tur og jeg elsker å spise frukt.')
finn_mest_brukte_ord('Biler er kule, tog er også kule')


```
