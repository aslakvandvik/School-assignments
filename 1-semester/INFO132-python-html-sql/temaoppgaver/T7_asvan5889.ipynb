# Oppgave 1

```python
# Oppgave 1
import tkinter as tk

def farvel_klikk():
    rotvindu.destroy()

rotvindu = tk.Tk()
rotvindu.title("Farvel Applikasjon")

farvel_knapp = tk.Button(rotvindu, text="Farvel", command=farvel_klikk)
farvel_knapp.pack(padx=20, pady=20)

rotvindu.mainloop()
```

# Oppgave 2

```python
# Oppgave 2
import tkinter as tk

def telleknapp():
    global teller
    teller += 1
    oppdater_knapp()

def oppdater_knapp():
    knapp["text"] = str(teller)

teller = 0

hovedvindu = tk.Tk()
hovedvindu.title("Inkrementer Applikasjon")

knapp = tk.Button(hovedvindu, text=str(teller), command=telleknapp)
knapp.pack(padx=20, pady=20)

hovedvindu.mainloop()
```
