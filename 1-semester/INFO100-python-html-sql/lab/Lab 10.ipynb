```python
# Oppgave 2. 

def hent_snittkarakterer(karakterer):
    karakterer_sum = 0
    for i in karakterer:
        karakterer_sum += i[1]
    return karakterer_sum/len(karakterer)


print(hent_snittkarakterer([('Alice', 3), ('Bob', 4), ('Charlie', 5)]))
print(hent_snittkarakterer([('Alice', 2), ('Bob', 3), ('Charlie', 4), ('Eve', 5)]))
```

```python
# Oppgave 3.

def flett_tuppel(tuple1, tuple2):
    resultat = []
    for i in range(max(len(tuple1), len(tuple2))):
        if i < len(tuple1):
            resultat.append(tuple1[i])
        if i < len(tuple2):
            resultat.append(tuple2[i])
    return tuple(resultat)

print(flett_tuppel((1, 2, 3), (4, 5, 6)))
print(flett_tuppel((1, 2, 3), ('a', 'b', 'c', 'd', 'e')))
```
