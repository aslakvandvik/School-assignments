# __Iterasjon__ – gjentatt utførelse (__løkke__, "__loop__")

* Ofte nyttig å gjenta programsetninger flere ganger
* Eksempel: finn neste skuddår

```python
def er_skuddår(årstall):
    """Skriv ut om årstallet er et skuddår."""  # fra F04
    if ((årstall % 4 == 0 and årstall % 100 != 0)
            or (årstall % 4 == 0 and årstall % 400 == 0)):
        print(f'{årstall} er et skuddår!')
    else:
        print(f'{årstall} er ikke et skuddår.')
```

```python
er_skuddår(2021)
```

```python
er_skuddår(2022)
```

```python
er_skuddår(2023)
```

```python
er_skuddår(2024)
```

```python
år = 2023
while not er_skuddår(år):
    år = år + 1
print('Neste skuddår er', år)
```

# `while`-setninger

![](F05/while-setning.png)

* Blokken repeteres så lenge betingelsen er oppfylt (`True`)
* Betingelsen (re-)evalueres _før_ hver runde 
  * og vil typisk inneholde variable som endres i blokken

* Løkkesetningene `for` og `while` automatiserer repetisjon
  * vi sier at programsetningene __itereres__, eller utføres i __løkke__ (__"loop"__)
  * hver enkelt repetisjon kalles en _runde_ i løkken, eller en _iterasjon_ (_"iteration"_)
* Programflyt ("flow of execution"):

1. `år = 2023`
1. `er_skuddår(år)`
1. `    # ...funksjonskroppen utføres...`
1. `    return False`
1. `not er_skuddår(år)` [...er nå True...]
1. `    år = år + 1`
1. `er_skuddår(år)`
1. `    # ...funksjonskroppen utføres med ny parameterverdi...`
1. `    return True`
1. `not er_skuddår(år)` [...er nå False...]
1. `print('Neste skuddår er', år)`

* Løkker kan også brukes i funksjoner:

```python
def neste_skuddår(år):
    while not er_skuddår(år):
        år += 1
    return år
```

```python
neste_skuddår(2023)
```

# Eksempel: `6*7*8*9*10*11 = ?`
* Produktet av tallene i et _lukket_ heltallsintervall \[m, n\]: `m * ... * n`

<!--
def intervallprodukt(fra, til):
    produkt = 1  # akkumulator
    teller = fra  # teller
    while teller < til:
        produkt = produkt*teller
        teller = teller+1

```python
def intervallprodukt(fra_og_med, til_og_med):
    produkt = 1  # akkumulator
    teller = fra_og_med  # teller
    while teller <= til_og_med:
        produkt *= teller
        teller += 1
    return produkt

print('Produktet er', intervallprodukt(2, 5))
```

# Eksempel: `6*8*10 = ?`
* Produktet av _partallene_ i et intervall: `m * ... * n`

```python
def intervallprodukt(fra_og_med, til_og_med):
    produkt = 1  # akkumulator
    teller = fra_og_med  # teller
    while teller <= til_og_med:
        if teller%2 == 0:  # eller: er_partall(teller)
            produkt *= teller
        teller += 1
    return produkt

print('Produktet er', intervallprodukt(2, 5))
```

* Mer avansert løsning

```python
def partallprodukt(fra_og_med, til_og_med):
    produkt = 1  # akkumulator
    teller = (fra_og_med 
                  if fra_og_med%2==0 
                  else fra_og_med+1)  # garanterer at teller er et partall
    while teller <= til_og_med:
        produkt = produkt*teller  # nå slipper vi if-testen
        teller = teller+2  # vi går opp i sprang på 2
    return produkt

print('Produktet er', partallprodukt(2, 5))
```

# Uendelige løkker

* Om betingelsen alltid evalueres til `True` fortsetter iterasjonene _«i det uendelige»_ (_"infinitely"_)

* For å unngå uendelig løkke:

# `break`

* Bryter ut av en løkke
  * ut av _hele løkken_
* Programflyten fortsetter fra _første programsetning etter løkken_
* `while True` i kombinasjon med `break` kan av og til gjøre koden mer lesbar

```python
handleliste = []

vare = input('Hva vil du handle? ')
while vare != 'ingenting':
    handleliste += [vare]
    vare = input('Hva mer vil du handle? ')

print('Handlelisten er:', handleliste)
```

```python
handleliste = []

while True:
    vare = input('Hva vil du handle? ')
    if vare == 'ingenting':
        break
    handleliste += [vare]

print('Handlelisten er:', handleliste)
```

# `continue`

* Bryter ut av én enkelt iterasjon (runde) i en løkke
  * bare ut av den ene _iterasjonen / runden_
* Programflyten fortsetter deretter med _neste runde_ i løkken

```python
handleliste = []

while True:
    vare = input('Hva vil du handle? ')
    if vare == 'ingenting':
        break
    if vare in ['øl', 'vin', 'sprit']:
        print('Vi skal ikke handle alkohol i dag!')
        continue
    handleliste += [vare]

print('Handlelisten er:', handleliste)
```

```python
lag = ['Brann', 'Lillestrøm', 'Rosenborg', 'Strømsgodset', 'Vålerenga']

# skriv ut alle seriekamper
for hjemmelag in lag:
    for bortelag in lag:
        if hjemmelag != bortelag:
            print('Kampen er', hjemmelag, '-', bortelag)
```
