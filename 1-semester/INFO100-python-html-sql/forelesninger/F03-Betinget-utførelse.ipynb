# Hva vet vi om Python-kode?
* Grunnleggende Python-begreper så langt:
  * __verdi__
    * f eks `12`, `3.14159`, `'INFO123'`, `True`, `False`, `None`...
    * har __type__ (f eks tall, tegnstrenger, sannhetsverdier/logiske verdier)
  * __variabel__
    * midlertidig navn på en verdi: `d`, `pi`, `emnekode`...
  * __uttrykk__ ("expression")
    * består av __operatorer__ og __operander__: `d * pi`)
    * operander: verdier, variable, deluttrykk eller funksjonskall
  * __funksjonskall__ (__argumenter__ og __opsjoner__)
    * argumenter og opsjoner
    * argumentene kan være skrevet som:
      * verdier, variable, uttrykk eller funksjonskall
    * __resultatverdi__ eller __sideeffekt__
  * __variabeltilordning__ ("variable assignment")
    * `o = d * pi`
    * _fra_ verdi, variabel, uttrykk, funksjonskall
    * _til_ variabel
  * __aritmetiske operatorer__
    * `+`, `-`, `*`, `/`, `//`, `%`, `**`
    * fra to tall til ett nytt tall
      * ...men flere av dem er _overbelastet_

# Betinget utførelse

* Blokken utføres kun hvis betingelsen evalueres til `True`

````python
alder = int(input('Hvor gammel er du? '))
myndighet = 'myndig'
if alder<18:
    myndighet = 'ikke myndig'  # merk innrykk (indentering) med 4 blanke tegn
print('Du er', myndighet)
````

# Alternativ uførelse
* `if` – `else` -setning

<img src="F03/if-else-setning.png" width=300/>

* Hvis betingelsen gir `True` utføres Blokk 1 (`if`-grenen)
* Ellers utføres Blokk 2 (`else`-grenen).

````python
alder = int(input('Hvor gammel er du? '))
if alder<18:
    print('Du er ikke myndig.')
else:
    print('Du er myndig.')
````

# Betinget alternativ utførelse
* `if` – `elif` – `else` -setning

<img src="F03/if-elif-else-setning.png" width=500/>

# Forenkling

* Nøsting og komplekse `if`-setninger kan ofte forenkles
* Eksempel fra tidligere:

````python
årstall = int(input('Oppgi et årstall:'))
if årstall % 4 == 0:
    if årstall % 100 == 0:
        if årstall % 400 == 0:
            print('Skuddår!')
        else:
            print('Ikke skuddår')
    else:
        print('Skuddår!')
else:
    print('Ikke skuddår')
````

* Vi kan forenkle blokkstrukturen
  * ingen nøsting
  * men mer komplekse logiske uttrykk
* F eks:

````python
årstall = int(input('Oppgi et årstall:'))
if ((årstall % 4 == 0 and årstall % 100 != 0)
        or (årstall % 4 == 0 and årstall % 400 == 0)):
    print('Skuddår!')
else:
    print('Ikke skuddår')
````

* Betingelsen (det logiske uttrykket) under er _ekvivalent_ med det over:

````python
årstall = int(input('Oppgi et årstall:'))
if årstall % 4 == 0 and (årstall % 100 != 0 or årstall % 400 == 0):
    print('Skuddår!')
else:
    print('Ikke skuddår')
````

# Ekvivalente uttrykk
* To uttrykk er __ekvivalente__ dersom de _alltid_ har samme (ekvivalente) verdi
  * «alltid» vil her si uansett variabelverdier og resultatverdier fra funksjonskall
  * for eksempel _distribusjon_ og _DeMorgans lover_

# Distribusjon
* `and` er _distributiv_ over `or`
  * og `or` er _distributiv_ over `and` 
* Gir mulighet for flere forenklinger, f eks<br/>
  `((årstall % 4 == 0 and årstall % 100 != 0) or (årstall % 4 == 0 and årstall % 400 == 0))`<br/>
  `    == (årstall % 4 == 0 and (årstall % 100 != 0 or årstall % 400 == 0))`
  

````python
alder = 35
not (alder<18 or alder>=67)
````

````python
(not alder<18) and (not alder>=67)
````

````python
alder>=18 and alder<67
````

# DeMorgans lover
* Sannhetsuttrykket<br/>
  `(not (alder<20 or alder>40))`<br/>
  `    == (alder>=20 and alder<=40)`
  * er _alltid sant_ (dvs uavhengig av verdien på `alder`)
  * på grunn av DeMorgans første lov!
* ...men _ikke glem parentesene!_

# DeMorgans lover
* DeMorgans (første) lov
  * `(not (logisk_uttrykk_1 or logisk_uttrykk_2))`<br/>
    `    == ((not logisk_uttrykk_1) and (not logisk_uttrykk_2))`
* DeMorgans (andre) lov
  * `(not (logisk_uttrykk_1 and logisk_uttrykk_2))`<br/>
    `    == ((not logisk_uttrykk_1) or (not logisk_uttrykk_2))`

# Eksempel: bonuslønn
* Du tjener 2000 per dag i _grunnlønn_
* Hvis du selger inntil 1000 enheter får du også 0.1 i _provisjon_ per enhet
* Men selger du to 1000 enheter eller mer får du i stedet 0.2 i _bonus_ per enhet

````python
grunnlønn = 2000  # per dag
provisjon = 0.1
bonus = 0.2

antall = int(input('Antall solgte enheter i dag: '))
påslag = (antall*provisjon) if antall<1000 else (antall*bonus)  # parentesene trengs ikke fordi det betingede uttrykket har lav presedens
dagslønn = grunnlønn + påslag
print('Du tjente', dagslønn, 'i dag.')
````

````python
grunnlønn = 2000  # per dag
provisjon = 0.1
bonus = 0.2

antall = int(input('Antall solgte enheter i dag: '))
påslag = antall * (provisjon if antall<1000 else bonus)  # her er parentesene nødvendige, ellers ville 'antall*provisjon' blitt utført først
dagslønn = grunnlønn + påslag
print('Du tjente', dagslønn, 'i dag.')
````

````python
grunnlønn = 2000  # per dag
provisjon = 0.1
bonus = 0.2

antall = int(input('Antall solgte enheter i dag: '))
if antall<1000:  # vi kunne også brukt vanlig betinget utførelse
    påslag = antall*provisjon
else:
    påslag = antall*bonus
dagslønn = grunnlønn + påslag
print('Du tjente', dagslønn, 'i dag.')
````
