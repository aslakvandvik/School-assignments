# HyperText Transfer Protocol (HTTP)

* HTTP (Hypertext Transfer Protocol) er kanskje den mest brukt _protokollen_ for å overføre informasjon på internettet/veven
* HTTP er en _asymmetrisk_ _forespørsel-svar_ (_"request-response"_) _klient-tjener_ (_"client-server"_) protokoll, som vist under
    - en HTTP klient sender en _forespørsel_ (_"request message"_) til en HTTP tjener
    - tjeneren returnerer et _svar_ (_"response message"_)
    - HTTP er med andre ord en _"pull"_-protokoll:
      - klienten _trekker ut_ informasjon fra tjeneren,
      - i stedet for at tjeneren _forer_ klienten med informasjon (_"push"_)

![](F12/HTTP.png)

* HTTP er _tilstandsløs_ (_"stateless"_):
  * hver forespørsel kan besvares uten at tjeneren må huske på tidligere forespørsler og svar
* HTTP støtter _forhandlinger_ om medietyper og formater:
  * klienter og tjenere kan dermed programmeres mer uavhengig av hverandre

_"The Hypertext Transfer Protocol (HTTP) is an application-level protocol for distributed, collaborative, hypermedia information systems.
It is a generic, stateless, protocol which can be used for many tasks beyond its use for hypertext, such as name servers and distributed object management systems, through extension of its request methods, error codes and headers."_ (RFC2616)

# Uniform Resource Locator (URL)

En URL (Uniform Resource Locator) brukes til å identifisere en ressurs på nettet entydig. URL har følgende syntaks:
* `protokoll://vertsnavn:port/sti-og-filnavn`

En URL har 4 deler:
- _protokoll_: applikasjonsnivåprotokollen som brukes av klient og tjener, f.eks., `HTTP`, `FTP` eller `SMTP`
- _vertsnavn_: DNS-domenenavnet (f.eks., `www.nowhere123.com`) eller IP-adressen (f.eks., 192.128.1.2) til tjenermaskinen (_"server"_)
- _port_: TCP-portnummeret som tjeneren lytter etter innkommende forespørsler fra klientene (_"client"_)
- _sti-og-filnavn_: Navnet og plasseringen av den forespurte ressursen, under tjenerens dokumentbasekatalog

Eksempel URL: `http://www.nowhere123.com/docs/index.html`
- _kommunikasjonsprotokollen_ er HTTP
- _vertsnavnet_ er `www.nowhere123.com`
- _portnummeret_ er ikke spesifisert i URL-en, og tar derfor standardnummeret, som er TCP-port `80` for HTTP
- _stien_ og _filnavnet_ for ressursen som skal lokaliseres er `/docs/` og `index.html`

Andre eksempler på URL-er er:

```
ftp://www.ftp.org/docs/test.txt
mailto:user@test101.com
news:soc.culture.Singapore
telnet://www.nowhere123.com/
```

````python
import urllib.request

url_str = 'http://www.uib.no/emne/INFO132'
respons = urllib.request.urlopen(url_str)
````

# JSON-data lagret i fil

* `json.load()` og `json.dump()`  # leser og skriver JSON-data fra og til filer

# Vevtjenester med argumenter
* GeoNames er en stor geografisk database på nettet
  * _http://www.geonames.org_
  * tilbyr også et vev-API
    * _http://api.geonames.org_

* HTTP protokollen støtter URL-er med egen argumentsyntaks
  * _http://api.geonames.org/findNearbyJSON?lat=60.390990&lng=5.320557&username=demo_
  * litt som et _hypotetisk_ Python-kall
    * `geonames.find_nearby_json(lat=60.390990, lng=5.320557, username='demo')`
    * ...men det finnes ikke noen slik standard Python-modul og -metode
  * dokumentasjon
    * _https://www.geonames.org/export/ws-overview.html_
  * registrering
    * _http://www.geonames.org/login_

````python
url_str = 'http://api.geonames.org/findNearbyJSON?' + \
          'lat=60.390990&lng=5.320557&username=demo'
respons = urllib.request.urlopen(url_str)
````

````python
respons.status, respons.msg
````

````python
print(respons.read().decode())  # 'demo'-kontoen overskrider veldig fort den daglige kvoten
````

* _Tips:_ legg parametrene i en fortegnelse

````python
parametre = { 
    'lat': 60.390990, 
    'lng': 5.320557, 
    'username': 'demo', 
}
````

  * og bruk `urlencode`-funksjonen

````python
tjenesteurl = 'http://api.geonames.org/findNearbyJSON?'
url_str = tjenesteurl + urllib.parse.urlencode(parametre)
````

````python
print(url_str)
````
