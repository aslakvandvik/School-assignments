# Hva vet vi om funksjoner?

* En __funksjon__ er en navngitt del av et program som produserer et bestemt __resultat__ og/eller en bestemt __effekt__
  * __resultatfunksjoner__ (også kalt __verdifunksjoner__) har bare _resultatverdi_:<br/>
    `type()`, `max()`, `min()`, `int()`, `float()`, `str()`, `bool()`, `round()`, `len()`, `dir()`...
  * __tomme ("void") funksjoner__ (også kalt __prosedyrer__) har bare _sideeffekt_:<br/>
    `print()`, `help()`...
  * noen funksjoner har _både_ sideeffekt og resultatverdi:<br/>
    `input()`...
* Resultatet/effekten kan avhenge av data vi kaller __argumenter__
  * argumenter kan være __navngitte__ og __valgfri__ fordi de har __standardverdi__
  * valgfri argumenter kalles også _opsjoner_
* Funksjoner utføres vha __funksjonskall__
  * `funksjonsnavn(argument_1, ..., argument_n)`
* Så lang har vi brukt de __innebygde funksjonene__ ("built-ins") i Python
  * ...snart skal vi definere våre egne!

````python
def partall(tall):
    """Sjekk om tallet er et partall."""
    resultat = (tall%2 == 0)
    return resultat  # dette er en resultatfunksjon!
````

# Navnerom ("name space")
* Python-tolken husker variabelverdiene våre fordi
  * instruksjonene kjører i det samme __navnerommet__
  * navnerommet holder orden på verdiene på de navngitte variablene våre
* Python-tolken og programmene vi skriver starter i det __globale navnerommet__
  * felles navnerom for hele programmet (som kan være Python-tolken)
  * __globale variable__ er tilgjengelige overalt mens programmet kjører
    * også inne i funksjonene
    * `globals()`  # lister de globale variablene og deres verdier
* Hver funksjon starter i et nytt __lokalt navnerom__
  * __lokale variable__ er kun tilgjengelige mens funksjonen utføres
    * `locals()`  # lister de lokale variablene og deres verdier
  * når funksjonen returnerer _blir dens lokale navnerom borte for godt_

````python
def skriv_dobbel_x_global():
    y = 2*x  # x er en global variabel, y er lokal
    print(y)
````

````python
x, y = 5, 0
skriv_dobbel_x_global()
print(x, y)
````

# Programflyt ("flow of execution")
* Et program er skrevet som en _sekvens_ av _programsetninger_ (_instruksjoner_)
* I utgangspunktet følger kjøring/eksekvering av programmet denne sekvensen («ovenfra og ned», "top down")
* Noen programsetninger kan endre rekkefølgen
  * betinget utførelse (valg, "conditionals"): `if ... elif ... else`
  * iterasjon (løkker, "loops" &rarr; F05): `while ...`, `for ...`
  * unntak ("exceptions"): `try ... except ...`
  * funksjonskall: `funksjonsnavn(...)`
  * og flere...
* __Programflyten__ er _rekkefølgen programsetningene faktisk blir utført i_
  * kan variere for hver kjøring/eksekvering

# Eksempel
* _Programmet_ er en sekvens av programsetninger

````python
"""Skriver ut om x er oddetall eller partall."""

def partall(tall):
    resultat = (tall%2 == 0)
    return resultat

def oddetall(tall):
    return not partall(tall)

x = 2
resultat = oddetall(x)
print(resultat)
````

1. ```x = 2```

1. ```    def partall(tall):```
1. ```        resultat = (tall%2 == 0)```
1. ```        return resultat```
1. ```    return not partall(tall)```
1. ```resultat = oddetall(x)```
1. ```print(resultat)```

# Programflyt og funksjonskall
* Dette skjer når programflyten utfører et funksjonskall (fra hovedprogrammet)
  1. eksekveringen av hovedprogrammet settes på pause (den suspenderes)
  2. kallets argumentverdier evalueres (hvis argumentene er uttrykk)
  3. et nytt lokalt navnerom lages for funksjonen
  4. funksjonsparametrene blir lokale variable i det nye, lokale navnerommet
  5. disse lokale variablene tilordnes argumentverdiene fra funksjonskallet
  6. funksjonens kropp utføres i det lokale navnerommet
  7. når funksjonskroppen er ferdig slettes variablene i det lokale navnerommet
  8. bare returverdien (som kan være `None`) sendes tilbake
  9. hovedprogrammet mottar returverdien
  10. den opprinnelige eksekveringen av hovedprogrammet (som var satt på pause) gjenopptas 

# Skop: variable har rekkevidde
* En funksjonsdefinisjon kan opprette lokale variable
  * parametervariable
    * definert i parentes i funksjonshodet
    * får sine verdier
      * _enten_ fra argumenter i funksjonskallet
      * _eller_ fra standardverdier i funksjonsdefinisjonen
  * variable opprettet på vanlig måte inne i funksjonskroppen
    * definert og får sine verdier ved tilordning:
    * f eks `resultat = (tall%2 == 0)`

* Funksjonens kropp utgjør __skopet__ ("scope") til dens lokale variable (f eks `tall`, `resultat`)
  * utenfor sitt skop har variabelnavn ingen mening (eller verdi)
  * _en lokal variabel i ett navnerom er uavhengig av andre variable med samme navn i andre navnerom_
* Variable opprettet utenfor funksjonene er __globale__
  * de har globalt skop: programmet i sin helhet
  * det globale skopet _omslutter_ alle andre skop

````python
import math
pi = math.pi  # pi har globalt skop

def pluss_pi(y):
    """Skriver ut tallet pluss pi."""
    sum_ = y+pi  # sum og y har lokalt skop, pi's skop er fremdeles globalt
    print(sum_)
          
def ganger_pi(z):
    """Skriver ut tallet ganger pi."""
    produkt = z*pi  # produkt og z har lokalt skop, pi's skop er fremdeles globalt
    print(produkt)
````

````python
def pluss_pi(y):
    """Skriver ut tallet pluss pi."""
    sum_ = y+pi
    print(sum_)

def ganger_pi(y):  # funksjoner i samme program kan ha samme parameternavn
    """Skriver ut tallet ganger pi."""
    produkt = y*pi  # y i ganger_pi-skopet er forskjellig fra y i pluss_pi
    print(produkt)
````

# Skopregelen
* En variabel kan ikke brukes/refereres til (er ikke __tilgjengelig__, __"accessible"__) utenfor dens skop
  * men en funksjonskropp kan bruke/referere til globale variable
* Med andre ord
  * lokale variable er «usynlige» ("invisible") utenfor funksjonen
  * globale variable er synlige overalt i programmet ("visibility")
    * ...men inne i en funksjon kan de være __dekket__ (eller _gjemt_, _"hidden"_) av en lokal variabel med samme navn
    * ...den lokale variabelen tar __presedens over__ (__"precedence over"__) eventuelle andre variable med samme navn i __omsluttende skop__ (__"encompassing scope"__)
* Det er skopregelen som bestemmer hva det lokale navnerommet til en funksjon inneholder

# Omsluttende skop: `global`
* En funksjon står fritt til å lese (bruke verdien til) variable i det globale navnerom
* Men: forsøk på å oppdatere (endre verdien til) en global variabel vil i stedet føre til at det opprettes en _ny lokal variabel_ med samme navn
  * `global x` sier at variabelen `x` likevel skal oppdateres i det globale navnerommet

````python
import math
pi = math.pi  # pi har globalt skop

def pluss_pi(y):
    """Skriver ut tallet pluss pi."""
    sum_ = y+pi  # sum og y har lokalt skop, pi fremdeles globalt
    print(sum_)

pluss_pi(2.0)
    
print(sum_)  # sum_ eksisterer kun i pluss_pi's lokale navnerom
````

````python
import math
pi = math.pi  # pi har globalt skop

def pluss_pi(y):
    """Skriver ut tallet pluss pi."""
    global sum_  # sum_ blir definert i det globale navnerommet
    sum_ = y+pi  # sum_ har globalt skop, y har lokalt skop, og pi er fremdeles global
    print(sum_)

pluss_pi(2.0)
    
print(sum_)
````

# Nøstede funksjoner
* En funksjon kan defineres innen i en annen funksjon

````python
def partall(tall):
    """Sjekk om tallet er et partall."""
    resultat = (tall%2 == 0)
    return resultat

x = 2
partall(x)
````

* Versjon med _indre funksjoner_:

````python
def partall(tall):
    """Sjekk om tallet er et partall."""

    def modulus2(tall):
        """Finn rest etter at tall er delt på to.""" 
        resultat = tall%2
        return resultat

    def er_null(tall):
        """Sjekk om tallet er null."""
        return tall==0

    rest = modulus2(tall)
    resultat = er_null(rest)
    return resultat

x = 2
partall(x)
````

* Mer kompakt versjon med _indre funksjoner_:

````python
def partall(tall):
    """Sjekk om tallet er et partall."""

    def modulus2(tall): return tall%2  # finn rest etter at tall er delt på to
    def er_null(tall): return tall==0  # sjekk om tallet er null

    return er_null(modulus2(tall))

x = 2
partall(x)  # x er et argument
````

* Kroppen til `partall` utgjør skopet til de lokale funksjonene `er_null` og `modulus2`
  * `partall`-navnerommet __omslutter__ (__"encompasses"__) `er_null`- og `modulus2`-navnerommene
