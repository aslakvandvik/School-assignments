# Hendelsesdrevet programmering<br/>("Event-driven programming")

* GUI-programmet styres via __GUI-elementer__
  * også kalt __skjermelementer__, __vinduselementer__, eller __"widgets"__
* Elementene er plassert i ett (eller flere) __vindu(er)__
  * __rotvinduet__ (__"root window"__) går i _kontinuerlig løkke_
    * __hendelsesløkke__ (__"event loop"__)
    * løkken _overvåker_ og _responderer på_ hendelser i vinduet
  * hendelser kan være f eks 
    * inndata i tastefelt, klikk på knapper, menyvalg, museoperasjoner (pek/klikk/...)
    * hendelsene behandles av __håndteringsfunksjoner og -metoder__
      * kalles __håndterere__ (__"handlers"__)

# Struktur for GUI-programmer
* Brukergrensesnittet har et _rotvindu_ som må opprettes som et `tkinter`-objekt
  * `rotvindu = tkinter.Tk()`  # oppretter rot-vinduet
* Den uendelige _hendelsesløkken_ startes vha metoden 
  * `rotvindu.mainloop()`  # starter hendelsesløkken
* ...og kan _avsluttes_ med metoden 
  * `rotvindu.destroy()`  # lukker og sletter rot-vinduet
* Dette skaper et dilemma:
  * hvordan skal vi få kalt denne metoden hvis hendelsesløkken (`.mainloop()`) er uendelig?  

* Eksempel på programstruktur:

```python
import tkinter

# Programkode for det programmet skal handle om (“modell”)

# Programkode som håndterer hendelser (“kontroll”)

rotvindu = tkinter.Tk()

# Programkode for GUI-elementer (“presentasjon”):
#   - definisjon av knapper, tekstfelt, menyer, andre vinduer, osv
#   - hvilke håndterere som tar seg av hvilke hendelser

# Eventuell programkode som starter hovedprogrammet

print('Før hendelsesløkken')
rotvindu.mainloop()
print('Etter hendelsesløkken')

# Eventuell programkode som avslutter hovedprogrammet

rotvindu.destroy()  # dette gir feilmelding, fordi rotvinduet er destruert
```

# Elementtypen `Button` - legg til en knapp

```python
from tkinter import Tk, Label, Button, RAISED, GROOVE

rotvindu = Tk()

hei = Label(rotvindu, text='Hei', width=10, bd=8, relief=GROOVE, 
            fg='red', bg='yellow')
hei.grid(row=0, column=0)

def lukk_vindu():
    print('Du har trykket på lukkeknappen!')
    rotvindu.destroy()  # .destroy()-metoden lukker vinduet

knapp = Button(rotvindu, text='lukk', command=lukk_vindu)  # merk: 'lukk_vindu' må skrives uten parenteser her!
knapp.grid(row=1, column=0)

rotvindu.mainloop()
```

* Generelt:
  * __elementene ("elements")__ blir åsted for __hendelser ("events")__ som er knyttet til __håndterere ("handlers")__
* Spesifikt:
  * __elementet "knapp"__ er åsted for __hendelsen "klikk"__ som er knyttet til __håndtereren ".destroy()"__
* Vi sier også at håndtereren er en __tilbakekallsfunksjon__ eller __-metode__ (__"callback function__ or __method"__)

# `element.config(...)`
* `element.config(opsjon=verdi, ...)`
  * oppdaterer _konfigurasjonen_ til et element
  * _etter_ at det er initialisert
* Konfigurasjonen vil si egenskaper som
  * størrelser, farver, rammer, osv
* Opsjonene til `element.config(...)` er stort sett de samme som opsjonene til `Elementtype()`-konstruktøren

```python
from tkinter import Tk, Label

rotvindu = Tk()

hei = Label(rotvindu, text='Hei')
hei.grid(row=0, column=0)

hei.config(text='Hallo', fg='blue')

rotvindu.mainloop()
```

```python
from tkinter import Tk, Label, Button

rotvindu = Tk()

hei = Label(rotvindu, text='Hei', width=10, fg='red', bg='yellow')
hei.grid(row=0, column=0)

knapp = Button(rotvindu, text='lukk', command=rotvindu.destroy)
knapp.grid(row=1, column=0)

# egendefinert håndteringsfunksjon
def skift_til_blått(): 
    hei.config(bg='blue')

skifteknapp = Button(rotvindu, text='blå', fg='blue', command=skift_til_blått)
skifteknapp.grid(row=1, column=1)

rotvindu.mainloop()
```

```python
from tkinter import Tk, Label, Button

rotvindu=Tk()

hei = Label(rotvindu, text='Hei', width=10,
            fg='blue', bg='yellow', font=('Arial', 12, 'bold italic'))
hei.grid(row=0, column=0)

knapp = Button(rotvindu, text='lukk', command=rotvindu.destroy)
knapp.grid(row=1, column=1)

farge = 'yellow'  # en vanlig global variabel

def skift_farge():
    global farge
    if farge == 'lightblue':
        farge = 'yellow'
    else:
        farge = 'lightblue'
    skifteknapp.config(bg=farge)

skifteknapp = Button(rotvindu, text='skift farge', bg=farge, command=skift_farge)
skifteknapp.grid(row=1, column=0)

rotvindu.mainloop()
```

# Automatisk oppdatering med variable
* Noen elementopsjoner kan navngis av en `tkinter`-variabel
  * disse er ikke vanlige Python-variable
  * når variabelverdien endres, blir elementet automatisk oppdatert
  * _uten at vi må kalle `.config()`-metoden_
* Oppretting av `tkinter`-variable
  * `var = tkinter.StringVar()`  # tegnstreng
  * `var = tkinter.IntVar()`  # heltall
  * `var = tkinter.DoubleVar()`  # desimaltall
  * `var = tkinter.BooleanVar()`  # `1/0` (tilsvarer `True`/`False`)
* Metoder
  * `var.set(verdi)`  # tilordning
  * `var.get()`  # returnerer verdien

# Elementtypen `Entry` – les inn data

```python
from tkinter import Tk, StringVar, Label, Button, Entry

rotvindu = Tk()

hilsen = StringVar()
hilsen.set('Hei')

hei = Label(rotvindu, textvariable=hilsen)
hei.grid(row=1, column=0)

def til_store_bokstaver():
    hilsen.set(hilsen.get().upper())

knapp = Button(rotvindu, text='Store bokstaver', command=til_store_bokstaver)
knapp.grid(row=1, column=1)

hilsenfelt = Entry(rotvindu, textvariable=hilsen, width=9)
hilsenfelt.grid(row=0, column=0)

rotvindu.mainloop()
```

# Elementtypen `Frame` - plassering av andre elementer

<img src="F11/frame.png" width=300/>

* Knappene 'Hei' og 'Hallo' hører sammen
  * men dynamikken i `.grid()` skaper uheldig layout
  * vi kan samle dem i en ramme (`Frame`)

```python
from tkinter import Tk, StringVar, Label, Frame, Button

rotvindu = Tk()

hilsen = StringVar()
hilsen.set('Hei')

hilsenfelt = Label(rotvindu, textvariable=hilsen, width=12)
hilsenfelt.grid(row=0, column=0)

knapperamme = Frame(rotvindu)
knapperamme.grid(row=1, column=0)

def hei(): hilsen.set('Hei')
def hallo(): hilsen.set('Hallo')

heiknapp = Button(knapperamme, text='Hei', command=hei)
heiknapp.grid(row=0, column=0)

halloknapp = Button(knapperamme, text='Hallo', command=hallo)
halloknapp.grid(row=0, column=1)

lukkeknapp = Button(rotvindu, text='Lukk', command=rotvindu.destroy)
lukkeknapp.grid(row=1, column=1)

rotvindu.mainloop()
```

# Mer om rammer og elementer

# Elementtypen `Radiobutton`

```python
from tkinter import Tk, StringVar, Label, Frame, SUNKEN, Radiobutton

rotvindu = Tk()

hilsen = StringVar()
hilsen.set('Hei')

hei = Label(rotvindu, width=10, textvariable=hilsen)
hei.grid(row=0, column=0)

hilsenramme = Frame(rotvindu, bd=2, relief=SUNKEN)
hilsenramme.grid(row=0, column=1)

halloknapp = Radiobutton(hilsenramme, text='Hallo', 
                         variable=hilsen, value='Hallo!')
halloknapp.grid(row=0,column=0)

heiknapp = Radiobutton(hilsenramme, text='Hei', 
                       variable=hilsen, value='Hei!')
heiknapp.grid(row=1,column=0)

rotvindu.mainloop()
```

# Elementtypen `Scale`

```python
from tkinter import Tk, IntVar, Label, Scale, HORIZONTAL

rotvindu = Tk()

skriftstørrelse = IntVar()
skriftstørrelse.set(9)

hei = Label(rotvindu, text='Hei',
            font=('Arial', int(skriftstørrelse.get()), 'bold italic'))
hei.grid(row=1, column=0)

def skift_skrift(_):  # '_' står for en parameter vi ikke bruker
    hei.config(font=('Arial', int(skriftstørrelse.get()), 'bold italic'))

skriftskala = Scale(rotvindu, variable=skriftstørrelse, command=skift_skrift, 
                    orient=HORIZONTAL, from_=9, to=30)
skriftskala.grid(row=0, column=0)

rotvindu.mainloop()
```

# Elementtypen `Menu`

```python
from tkinter import Tk, Label, Menu

rotvindu = Tk()

hei = Label(rotvindu, text='Hei')
hei.grid(row=0, column=0)

def lukk(): rotvindu.destroy()
def rød(): hei.config(fg='red')
def blå(): hei.config(fg='blue')
def grønn(): hei.config(fg='green')

menylinje = Menu(rotvindu)
meny = Menu(menylinje)
meny.add_command(label='Rød', command=rød)
meny.add_command(label='Blå', command=blå)
meny.add_command(label='Grønn', command=grønn)
menylinje.add_cascade(label='Farge', menu=meny)
menylinje.add_command(label='Lukk', command=lukk)

rotvindu.config(menu=menylinje)
rotvindu.mainloop()
```

# Elementtypen `TopLevel` – lag et nytt vindu

```python
from tkinter import Tk, StringVar, Entry, Toplevel, Label, Button

rotvindu = Tk()

hilsen = StringVar()
hilsen.set('Hei')

hilsenfelt = Entry(rotvindu, textvariable=hilsen, width=9)
hilsenfelt.grid(row=0, column=0)

def lag_vindu():
    nytt_vindu = Toplevel()
    hei = Label(nytt_vindu, text=hilsen.get())
    hei.grid(row=0, column=0)

knapp = Button(rotvindu, text='Nytt vindu', command=lag_vindu)
knapp.grid(row=1, column=0)

rotvindu.mainloop()
```
