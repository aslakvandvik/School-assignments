# Objekter
* Et __objekt__ består av 
  * _variable_ (kalt __attributter__) med 
  * tilhørende _operasjoner/funksjoner_ (kalt __metoder__) 
  * som til sammen utgjør en enhet
* Noen av de samme ideene som for funksjoner, men
  * objektene kan ha mange _metoder_ (funksjoner)
  * objektene har egne _attributter_ (variable) som er __persistente__ (__"persistent"__) mens programmet kjører
    * (mens lokale variable i en funksjon er _midlertidige_)
* ___Hvorfor?___
* Objekter er viktige
  * som _konstruksjonsmekanisme_ og
  * som _abstraksjonsmekanisme_

# Objekter som _konstruksjonsmekanisme_ 
* Objekter kan brukes til å danne sammensatte/komplekse data og datastrukturer
  * som i eksemplene ovenfor

# Objekter som _abstraksjonsmekanisme_ 
* Objekter kan brukes til å definere programstruktur
* Når vi programmerer har vi ofte mange objekter som ligner hverandre
  * f eks mange personobjekter
  * andre eksempler: studenter, emner, forelesere, studieretninger...
  * de har mye til felles som vi bare burde programmere en gang

* Navnekonvensjoner
  * `dette_er_navnet_til_en_variabel_eller_funksjon`
  * `dette_er_navnet_til_et_attributt_eller_en_metode`
  * `dette_er_navnet_til_et_objekt`
  * `DetteErNavnetTilEnKlasse`
  * men:
    * dette er (gode) _anbefalinger_ mer enn _regler_
    * mange _innebygde_ Python-klasser har korte navn med små bokstaver
    * `int`, `float`, `str`, `bool`, `list`, `type`, ...

# Metoder
* Så langt har klasser og objekter handlet mest om data_struktur_
  * vi har sagt lite om _oppførsel_
* Objekter kan også ha operasjoner, kalt __metoder__ (__"methods"__)
* Metoder ligner veldig funksjoner, men de
  * er definert inne i klassedefinisjoner
  * kalles ved hjelp av _prikknotasjon_, f eks
    * `handleliste.append('brød')`

  * så kan vi bruke _instansiering_ for å lage objekter for enkeltpersoner
  * vi slipper å duplisere kode
    * f eks _initialiserer_ vi bare attributtene én gang
  * &rarr; klasser og objekter som _konstruksjonsmekanisme_
* Vi kan samle alt som har med personer å gjøre på samme sted i programmet
  * når vi skal endre eller utvide er det lettere å finne igjen koden
  * det er lettere å se hvilke konsekvenser endringen/utvidelsen får

# `self`-parameteren
* En metode må ha objektet sitt som parameter
  * i metodekallet `kari.skriv_ut()` har `skriv_ut()` ett argument:
    * verdien av `kari`
  * i metodekallet `handleliste.append('brød')` har`append()` to argumenter:
    * verdien av `handleliste` og litteralet `'brød'`
* I metodedefinisjoner kommer denne parameteren alltid _først_
  * ...og den bør hete `self` (konvensjon)
  * `def metodenavn(self, param_1 , ... , param_n): ...`
  * f eks i definisjonen av klassen `Person`:
    * `def skriv_ut(self): ...`
  * f eks i definisjonen av (den innebygde) klassen `List`:
    * `def append(self, element): ...`

# __Initialisering__ (__"initialisation"__) av objekter
* Dersom en klasse definerer metoden `__init__(self, ...)` så blir den utført automatisk hver gang en ny instans opprettes
  * i _objektorientert programmering_ kalles dette en __konstruktør__ (__"constructor"__)

````python
class Person:
    """
    Holder orden på personopplysninger.
    """
    nasjonalitet = 'norsk'

    def __init__(self, fornavn, etternavn, telefon,
                 gateadresse, poststed, nasjonalitet=None):
        self.navn = Navn()
        self.navn.fornavn = fornavn
        self.navn.etternavn = etternavn
        self.telefon = telefon
        self.adresse = Adresse()
        self.adresse.gateadresse = gateadresse
        self.adresse.poststed = poststed
        if nasjonalitet:
            self.nasjonalitet = nasjonalitet
        else:
            self.nasjonalitet = Person.nasjonalitet
    
    def initialer(self):
        return self.navn.fornavn[0]+self.navn.etternavn[0]
    
    def skriv_ut(self):
        """
        Skriver ut en person.
        """
        print(self.navn.fornavn, self.navn.etternavn, 'er',
              self.nasjonalitet, 'og har telefonnummer', self.telefon,
              'og addresse', self.adresse.gateadresse, 'i', 
              self.adresse.poststed, end='.\n')
````

````python
kari = Person(
    fornavn='Kari',
    etternavn='Normann', 
    telefon=98765432, 
    gateadresse='Tullevei 9', 
    poststed='9876 Tøysedal',   
)
kari.skriv_ut()
````

````python
Person.nasjonalitet = 'finsk'
````

````python
kari.skriv_ut()
````

````python
per = Person('Per', 'Person', 99778891, 'Nedvei 5', '9855 Oppberg')
````

````python
per.skriv_ut()
````

# På tide med en innrømmelse...
* En funksjon er
  * _en navngitt del av et program som produserer et bestemt resultat og/eller effekt_
* Disse er _ekte funksjoner_ (importert fra __builtins__)
  * `print()`, `max()`, `min()`, `round()`, `input()`, `len()`, `globals()`, `dir()`, ...
* Disse er _klasser_ (også fra modulen __builtins__)
  * `type()`, `int()`, `float()`, `str()`, `bool()`, `list()`, ...
  * de er ikke ordentlige funksjoner
  * men _konstruktører_ som lager nye objekter av klassen
    * gjennom å kalle `__init__()`-metodene deres
* Python er så elegant laget at vi normalt ikke behøver å bry oss med forskjellen

````python
class Bankkonto:
    """
    Hold orden på bankkontoer.
    """
    kontonummer = None
    navn = None
    saldo = None
    rentesats = None
    neste_kontonummer = 10000
    
    def __init__(self, navn, saldo=0.0, rentesats=1.5):
        self.navn = navn
        self.saldo = saldo
        self.rentesats = rentesats
        self.kontonummer = Bankkonto.neste_kontonummer
        Bankkonto.neste_kontonummer += 1

    def innskudd(self, beløp):
        print('Setter inn', beløp, 'på konto nummer', self.kontonummer)
        self.saldo += beløp
    
    def uttak(self, beløp):
        if beløp <= self.saldo:
            print('Tar ut', beløp, 'fra konto nummer', self.kontonummer)
            self.saldo -= beløp
        else:
            print('Ikke dekning for', beløp, 'på konto nummer', self.kontonummer)
    
    def skriv_ut(self):
        print(self.navn, 'har', self.saldo, 'på konto nummer', 
              self.kontonummer, 'med rentesats', self.rentesats, end='.\n')

oles_konto = Bankkonto(navn='Ole Olsen', saldo=1000)
oles_konto.innskudd(1000)
oles_konto.uttak(2000)
oles_konto.skriv_ut()

jens_konto = Bankkonto('Jens Jensen', 45000, rentesats=3.0)
jens_konto.uttak(20000)
jens_konto.skriv_ut()
````

# Moduler
  * `math`, `random`, ... og _builtins_, ...
* En samling av elementer (variable, funksjoner, klasser)
  * lagret separat i en egen fil
  * ment for å `import`-eres inn i andre programmer (og moduler)
* Modulen er som en programfil, 
  * men vanligvis ikke er ment til å eksekveres.
* Nyttig til
  * gjenbruk av funksjoner
  * strukturering av større programmer ved å fordele dem på flere moduler

```
import navn_på_modul

...

navn_på_modul.variabelnavn

...

navn_på_modul.funksjonsnavn()
```

* _Prikknotasjonen_ brukes nå til å navngi
  * attributter og metoder i klasser og objekter
  * variable, funksjoner, klasser og objekter i moduler
* Objekter, klasser og moduler har sine egne navnerom
  * men prikknotasjonen _åpner skopet_ deres

# Alternative import-setninger
* Import av utvalgte modulelementer
  * `from modulnavn import variabel_a, funksjon_b, objekt_c`
  * `variabel_a`, `funksjon_b`, `objekt_c` kan nå navngis uten prikknotasjon
* Import av alle element i en modul (når det ikke er fare for navnekonflikt)
