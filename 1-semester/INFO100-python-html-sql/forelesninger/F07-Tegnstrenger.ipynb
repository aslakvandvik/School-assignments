# Segmentering ("slicing") av tegnstrenger (repetisjon)
* Segmentering gir tilgang til delstrenger av teksten:
  * `jj[0:4]`, `jj[-5:-1]`, `jj[::5]`
* En delstreng definert ved start- og sluttindeks
  * `tekst[m:n]`
  * utstnittet av tekst fra og med `m` og til men ikke med `n`
  * `tekst[m:m+1]` gir det samme som `tekst[m]`
* Utsnitt uten start- eller sluttindeks
  * `tekst[:n]` betyr `tekst[0:n]`
  * `tekst[m:]` betyr `tekst[m:len(tekst)]`
  * `tekst[:]` betyr `tekst[0:len(tekst)]`

```python
jj[0:4]
```

```python
jj[-5:-1]
```

```python
jj[-1]
```

```python
jj[-5:]  # siste 'Juba'
```

```python
jj[:4]  # første 'Juba'
```

```python
jj[ 0:10:2 ]
```

# Tegnstrenger er uforanderlige
* Verdien til et tekstobjekt kan ikke modifiseres (_muteres_)
  * __immutabilitet__ (__"immutability"__, _ikke-muterbarhet_)
* Det kan se ut som om`str`-metoder og operatorer _forandrer_ argumentene sine
  * men i virkeligheten _returnerer de nye tegnstrenger_ (`str`-objekter)

```python
jj1 = 'Juba Juba!'
jj2 = jj1[0:4]
jj3 = jj2 + ' Juba!'
jj1 == jj3  # sammenligningsoperatoren '==' tester om to verdier er like
```

* Variablene `jj1` og `jj3` inneholder samme tegnstrengsverdi, men de er ikke (nødvendigvis) _samme_ tegnstrengsobjekt

```python
jj1 is jj3  # sammenligningsoperatoren 'is' tester om to variabler inneholder samme objekt
```

* Dette er annerledes for lister:

```python
liste = ['brød', 'melk', 'smør', 'ost']
original_liste = liste
liste.remove('smør')
liste.remove('ost')
liste.append('smør')
liste.append('ost')
```

```python
original_liste == liste
```

```python
original_liste is liste
```

* Listen har vært det samme objektet hele tiden
  * men listeobjektet har inneholdt forskjellige elementer underveis
* Lister er _foranderlige_ eller _endrbare_ (_"immutable"_) 
* Dette går fint for lister:

```python
liste[2] = 'Bremykt'
```

* ...men ikke for tegnstrenger

```python
jj1[5] = 'j'
```

* Lag en funksjon `finn_alle(tekst, delstreng)` som finner startposisjonene til alle forekomster av delstrengen og returnerer disse i en liste

<!--

```python
def finn_alle(tekst, delstreng):
    liste = []
    pos = -1
    while True:
        pos = tekst.find(delstreng, pos+1)
        if pos==-1:
            break
        liste.append(pos)
    return liste
```

```python
finn_alle('Akkarkabaret', 'k')
```

# Formatering (lærebokstil)
* Teksten `t` settes inn for `%s`
  * `' %s ' % t`
* Heltallet `n` settes inn for `%d`
  * `' %d ' % n`
* Heltallet `n` skrives på totalt `v` plasser
  * `' %vd' % n`
* Kommatallet `n` skrives med `d` desimaler
  * `' %.df ' % n`
* Kommatallet `n` skrives på `v` plasser med `d` desimaler
  * `' %v.df ' % n`
* For å sette inn flere verdier i samme streng, må verdiene skrives i et _tuppel_ 
  * enda en _sekvenstype_ &rarr; F10
  * ligner på lister, men
    * skrives med vanlige parenteser `( ... )`
    * er _uforanderlig_ (_immuterbar_, _ikke-muterbar_)

```python
'xxx %.3f yyy' % 3.1415926
```

```python
antall = 5
enhet = 'dager'
snitt = 1.456
'Snittemperatur siste %d %s er %.3f mm' % (antall, enhet, snitt)
```

```python
antall = 3
enhet = 'uker'
snitt = 10.876
'Snittemperatur siste %d %s er %.3f mm' % (antall, enhet, snitt)
```

# Formatering (f-strenger)
* Teksten t settes inn for %s
  * `' %s ' % t`, `f' {t} ' `
* Heltallet n settes inn for %d
  * `' %d ' % n`, `f' {n} ' `
* Heltallet n skrives på totalt v _plasser_
  * `' %vd' % n`, `f' {n:v} '`
* Kommatallet n skrives med s _sifre_
  * `' %.df ' % n`, `f' {n:.s} '`
* Kommatallet n skrives på v plasser med s _sifre_
  * `' %v.df ' % n`, `f' {n:v.s} '`

```python
f'xxx {3.1415926:.3} yyy'
```
