# Del 1 - Lister

# Repetisjon av sekvenstyper
* Liste – sekvens av elementer
  * `[ 1, 2, 'tre', ['4', 'fem'] ]`
  * `[ ]`  (_tom liste_, "_empty list_")
  * elementene kan ha _vilkårlig type_
    * men vanligvis er de av _samme type_
* Tegnstreng – sekvens av tegn
  * jj = 'Juba Juba!'
  * `''`  # en tom streng
  * elementene er _tegn_ (`str`)
* Intervall ("range") – gir sekvens av tall
  * `range(m, n)` gir sekvensen `m, m+1, m+2, ... , n-1`
  * f eks `range(5, 10)` gir sekvensen `5, 6, 7, 8, 9`
  * elementene er _heltall_ (`int`)
  * _merk:_ slutttallet (`n`) er _ikke_ med i intervallet

# Indeksering og segmentering

* Positive og negative _indekser_ (_posisjoner_)
    * første posisjon er '0' og siste posisjon er '-1' eller 'len(sekvens)-1'
       
<img src="F09/negative-indekser.png" width=200/>

* Sekvenser kan
  * __indekseres__ ("__indexing__"): `jj[0]`, `jj[-1]`
    * tallet `i` i uttrykket `sekvens[i]` kalles _indeks_ (_"index"_)
  * __segmenteres__ ("__slicing__"): `jj[0:4]`, `jj[-5:-1]`, `jj[::5]`
    * `sekvens[fra_indeks : til_indeks : steg]`
    * _steg_ kalles _"stride"_ på engelsk
    * alle tre er _valgfri_ _(optional)_

# Operatorer, funksjoner og metoder

* Operatorene `*` og `+` kopierer og konkatenerer (slår sammen) lister og tegnstrenger (men ikke intervaller)
* Operatoren `e in S` sjekker om verdien `e` finnes som element i sekvensen `S`
* Funksjonen `len(sekvens)` gir antall elementer som før

# Lister er _foranderlige_ (_endrbare_, _muterbare_, _"mutable"_)

````python
def snu(liste):
    resultat = []
    for elem in liste:
        resultat = [elem] + resultat
    return resultat

# test
navn = ['Ola', 'Ana', 'Ali', 'Kari']
snu(navn)
````

````python
def snu(liste):
    resultat = []
    for pos in range(len(liste)-1, -1, -1):
        elem = liste[pos]
        resultat.append(elem)
    return resultat

# test
navn = ['Ola', 'Ana', 'Ali', 'Kari']
snu(navn)
````

````python
def snu(liste):
    return liste[::-1]

# test
navn = ['Ola', 'Ana', 'Ali', 'Kari']
snu(navn)
````

````python
def snu(liste):
    """Eksempel på rekursiv løsning"""
    if len(liste) <= 1:
        return liste
    else:
        snudd_liste = snu(liste[1:])
        snudd_liste.append(liste[0])
        return snudd_liste

# test
navn = ['Ola', 'Ana', 'Ali', 'Kari']
snu(navn)
````

# Fortegnelser ("dictionaries")
* Type `<class 'dict'>`
* Lagrer assosiasjoner mellom objekter i primærlager
* __Samlinger__ (__"collections"__) av __par__ på formen `nøkkel: verdi` ("`key: value`")

```
{  
    nøkkel_1: verdi_1 , 
    nøkkel_2: verdi_2 , 
    ... , 
    nøkkel_N: verdi_N , 
}
```

* Nøklene og verdiene kan ha _vilkårlige typer_
  * men det er vanlig at _alle nøklene har samme type_
  * og at _alle verdiene har samme type_
  * _merk:_ vi bruker ordet _verdi_ (_"value"_) på to ulike måter nå!

# Oppslag i fortegnelser ("dictionaries")

# `for`-løkke
* `for k in fortegnelse: ...`
  * traverserer nøklene i fortegnelsen _i innsettingsrekkefølge_
  * _siden i Python 3.7_

````python
for pos in resultatliste:
    print(pos)
````

````python
for pos in resultatliste:
    print(pos, resultatliste[pos])
````

# Fortegnelser minner om lister (på overflaten)

* Fortegnelser med heltallsnøkler kan dermed minne om lister på overflaten
  * men er _mer fleksible_
  * or men er _annerledes organisert_ internt
  * fortegnelser bruker _"hashing"_ av nøkkelverdier
* En liste kan _simuleres_ med en fortegnelse som bruker posisjon (listeindeks) som (heltalls-)nøkkel
  * kan være mer effektiv med mange elementer

# Fordeler med fortegnelser ("dictionaries")
* Kan angi verdier på flere måter enn ved posisjon/indeks (som er alltid er heltall)
* Kan angi verdier ved hjelp av kombinasjoner av flere verdier (skal vi se siden)
* Kan angi verdier _uavhengig av posisjon_
  * men fortegnelsen kan likevel traverseres med `for`-løkker
