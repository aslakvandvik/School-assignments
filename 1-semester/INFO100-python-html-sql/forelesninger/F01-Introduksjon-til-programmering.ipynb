# Hva er programmering?
* Å gi instruksjoner til en datamaskin ved hjelp av et programmeringsspråk
  * f eks programmeringssråket Python
* Resultatet er et __program__
  * et enklere program kan også kalles et __skript__
* Et program er en sekvens av __programsetninger__ (__instruksjoner__)
* Setningene i et program kalles også __programkoden__ eller __kildekoden__ ("source code")

# Eksempel:
![](F01/eksempelprogram.png)

Eksempelprogrammet
1. ber brukeren om et filnavn
1. leser innholdet filen
1. går gjennom hver linje i filen
1. går gjennom hvert ord i linjen
1. teller hvor mange ganger hvert ord forekommer i filen
1. finner hvilket ord som forekommer flest ganger
1. skriver ut ordet og hvor mange ganger det forekommer

```python
navn = input('Skriv inn filnavn:')
fil = open(navn)
ordteller = dict()

for linje in fil:
    ord_i_linjen = linje.split()
    for ord in ord_i_linjen:
        ordteller[ord] = ordteller.get(ord, 0) + 1
fil.close()

maksantall = None
maksord = None
for ord, antall in ordteller.items():
    if maksantall is None or antall > maksantall:
        maksantall = antall
        maksord = ord
        
print(maksord, maksantall)
```

# Tolkning av program ("Interpretation")
* Tolkning av programfil er et alternativ til kompilering
* Programsetningene skrives og leses inn én og én
* Tolken oversetter («interpreterer») dem «i farta» etter behov
  * uten at de må lagres i en programfil
  * uten (synlig) kompilering
* «Under panseret» gjør Python vanligvis en mellomting
  * programsetningene kan skrives inn én og én eller leses fra programfil
  * så blir de kompilert «i farta» til en mellomkode (__bytekode__)
  * så blir mellomkoden/bytekoden tolket
* Men fra utsiden _ser Python mest ut_ som det er tolket

# Lesbarhet
* Vær snill med leseren!
* Bruk meningsfulle variabelnavn og forklarende kommentarer
    * ikke slik: 

```python
x = 4
y = 450000+520000+498000+61000
z = y/x
print(y, 'fordelt på', x)
print('Gir i snitt', z)
```

  * men slik:

```python
# Program for å beregne snittlønn ved å summere alle
# ansattes lønninger og dele på antall ansatte.

# antall ansatte og deres sammenlagte lønn
# må oppdateres ved endringer i stab/lønninger
antall_ansatte = 4
sammenlagt_lønn = 450000+520000+498000+61000

# beregn snittlønn
snittlønn = sammenlagt_lønn/antall_ansatte

# skriv ut resultat
print(sammenlagt_lønn, 'i total lønn fordelt på', antall_ansatte, 'ansatte')
print('Gir i snitt', snittlønn, 'per ansatt')
```

* Dette eksemplet går faktisk litt over streken...
* For eksempel er kommentaren `# beregn snittlønn` antagelig unødvendig siden koden er så klar og enkel.

# Programmeringsfeil
* Syntaksfeil
  * ulovlig formulering
  * programmet avvises med feilmelding
    * _brudd på reglene om grammatikk og reserverte ord_
* Logiske feil
  * lovlig formulering, men ikke gjennomførbart
  * programmet avbrytes med feilmelding
    * riktig bruk av grammatikk og reserverte ord, men
    * _feil bruk av programmeringsspråket_
* Semantiske feil
  * lovlig formulering
  * programmet kjører som det skal uten feilmelding
  * har ikke ønsket resultat/virkning
    * riktig bruk av programmeringsspråket, men
    * _feil i problemforståelse eller -løsning_
