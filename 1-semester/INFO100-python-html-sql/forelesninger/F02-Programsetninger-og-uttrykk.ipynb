# Hva vet vi om Python-kode?
* Grunnleggende Python-begreper så langt:
  * __verdi__ ("value")
    * `12`, `3.14159`, `'INFO132'`
    * har __type__ (f eks heltall, flyttall og tegnstrenger)
  * __variabel__
    * midlertidig navn på en verdi: `d`, `pi`, `emnekode`
  * __uttrykk__ ("__expression__")
    * består av __operatorer__ og __operander__: `d * pi`
    * operandene kan være verdier, variable eller deluttrykk
  * __funksjonskall__
    * `print(emnekode, end=' ')`
    * __argumentene__ kan være
      * verdier, variable eller uttrykk
      * navngitte og valgfri
  * __variabeltilordning__ ("__assignment__")
    * `o = d * pi`
    * av verdi, variabel eller uttrykk

# Grunnleggende verdityper
* Kalles også bare _typer_...
* __Heltall__ ("__integer__", `int`)
* __Flyttall__ ("__floating-point number__", `float`)
* __Tegnstrenger__ ("__character string__"/"__string__", `str`)
  * sekvens av tegn (bokstaver, sifre, skilletegn)
  * den tomme strengen: `''`	eller	`“”`
* __Sannhetsverdier__, __logiske verdier__ ("__Boolean__", `bool`)
  * `True`, `False`
* __Ingen verdi__
  * `None`

````python
type(13), type(3.14159), type('Abrakadabra!'), type(True), type(False), type(None)
````

* I tillegg til de grunnleggende (_atomiske_) verditypene finnes _sammensatte_ typer
  * tegnstrenger (`str`) er egentlig sammensatte typer

# Eksempel
* Program som regner x antall minutter om til timer og minutter

````python
svar = input('Angi antall minutter: ')  # svar blir en tegnstreng
antall_minutter = int(svar)  # antall_minutter blir et heltall
timer = antall_minutter // 60
minutter = antall_minutter % 60
print(antall_minutter, 'm =', timer, 't +', minutter, 'm')
````

# Uttrykk («Expression»)
* _Programkode som beregner en verdi_
* Uttrykk formuleres ved hjelp av
  * litteraler (kommer snart!)
  * variable
  * operatorer
  * funksjoner (mer siden)
* Uttrykk kan være
  * atomiske
  * sammensatte

# Atomiske uttrykk<br/>("Atomic expressions")
* Består av enten 
  * ett enkelt litteral
  * en enkel variabel
  * ett enkelt funksjonskall
* Litteraler er _måter å skrive verdier på_
`5`, `53`, `3.14159`, `'Hei'`, `False`
* Variabler er navn på verdier (litt forenklet)
  * det er vi som bestemmer navnene på variabler
  * de er oftest midlertidige (men det finnes også __KONSTANTER__)
* Funksjonskall er navn på funksjoner
  * med argumenter og opsjoner i parentes
  * `min('2.18282', '3.14159')`

# Litteraler
* Måter å skrive enkeltverdier på i Python
  * heltall (`int`): `234`
  * desimaltall (`float`): `3.14159`
  * flyttall (`float`): `5.972e24`, `59.72e23`, altså `5.972 * 10**24`
    * så desimaltall er egentlig bare forenklede flyttall
  * tegnstrenger (`str`): `'Juba juba'`, `"Dum og deilig"`
    * Unicode-strenger: `u'\N{degree celsius}'` (`u` foran strengen trengs ikke i Python 3) 
    * format-strenger: `f'{timer} t + {minutter} m'`
  * logiske verdier (`bool`): `True`, `False`
* ...det finnes noen flere
  * ingen verdi ('NoneType'): `None`
  * flere typer tegnstrenger
  * visninger («displays») av sammensatte typer
* Samme verdi kan skrives med flere litteraler:
  * `1.`, `1.0`, `1.000000`, `1e0`
  * eller uttrykk:
    * `float(1)`, `0.5 + 0.5`
    * `float(True)  # _ikke_ gjør dette hjemme :-)`

````python
1 == 1. == 1.0 == 1.000 == 1e0 == 0.5+0.5 == float(True) == float(1) == int(1.000)
````

* ...men dette er en risikabel måte å programmere på...

````python
'\N{degree celsius}'
````

# Funksjoner
* En funksjon er en navngitt del av et program som produserer et bestemt resultat og/eller effekt
  * `print()`, `type()`, `max()`, `min()`, `int()`, `help()`, `float()`, `str()`, `bool()`,  ...
* Resultatet/effekten kan avhenge av inndata vi kaller argumenter
  * navngitte argumenter kan være valgfri ("optional")
* Funksjoner anvendes vha funksjonskall
  * `funksjonsnavn(argument_1, ..., argument_n)`
* Så lang har vi brukt innebygde funksjoner i Python
  * etter hvert skal vi lage våre egne
* Vi kan skille mellom to typer funksjoner:
  * __tomme__ («void») __funksjoner__ («prosedyrer») har nyttige effekter og returnerer ikke noe resultat
  * __resultatfunksjoner__ returnerer et nyttig resultat (en __funksjonsverdi__) og har minst mulig __sideeffekt__
  * de to bør _ikke_ blandes mer enn nødvendig

````python
x = print(5)  # print() er en tom funksjon
````

````python
x  # x er nå None og gir ikke synlig resultat som en instruksjon til tolken
````

````python
print(x)  # men x kan skrives ut med print()
````

````python
y = max(4, 2, 3, 1)  # max() er en resultatfunksjon som ikke gir synlig resultat
````

````python
y  # men y gir synlig resultat - uten print(y)
````

# Nyttig modul: random
* `import random`
* Stort utvalg funksjoner for å generere tilfeldige tall og verdier
  * `random.randint(tall_1, tall_2)` returnerer et tilfeldig heltall mellom `tall_1` og `tall_2`
