# Enkle tekstfiler
* En __tekstfil__ er en sekvens av tekstlinjer (eller tegn)
* Filoperasjoner:
  * åpne, lese, skrive, legge til, lukke ("open", "read", "write", "append", "close")
* Vanlig lese- og skriveoperasjoner er __sekvensielle__: 
  * de arbeider seg alltid framover i sekvensen: 
  * linje for linje (eller tegn for tegn)

# Åpne en fil
* Funksjonen `open(filnavn, mode='r')`:
  * gjør filen klar til bruk i et __modus__ (__"mode"__)
  * returnerer en __kobling__ (__"handle"__, eller __håndtak__) til filen
* Filtilkoblingen er et _objekt_ som tilbyr nyttige _metoder_
* Modus angir hvordan filen kan brukes
  * `open(filnavn)` og `open(filnavn, 'r')`:
    * "read" - åpner for lesing av eksisterende fil
  * `open(filnavn, 'w')`:
    * "write" - åpner for skriving av ny (eller overskriving av eksisterende) fil
  * `open(filnavn, 'a')`:
    * "append" - åpner for å legge til på slutten av eksisterende fil

# Leseoperasjoner
* `fil.readline()`
  * leser neste linje fra filen og returnerer den som en tegnstreng (med linjeskift `\n`)
  * returnerer `''` (tom streng) når det ikke er flere linjer å lese
* `fil.readlines()`
  * leser resten av filen og returnerer en liste over linjer (med linjeskift `\n`)
  * returnerer `[]` (tom streng) når det ikke er flere linjer å lese
  * __NB:__ _ikke_ gjør dette med store filer
* `fil.read(n)`
  * leser de neste `n` tegnene og returnerer dem som en tekst
* `fil.read()`
  * leser resten av filen og returnerer som en tekst.
  * __NB:__ _ikke_ gjør dette med store filer

# Automatisk lukking med `with`

* `with open(filnavn, mode='x') as file:`
  * inne i `with`-blokken kan vi bruke `file` som kobling til den åpne filen
  * `with`-blokken må indenteres
  * når vi går ut av `with`-blokken lukkes filen automatisk
* Samme som:
  * `file = open(filnavn, mode='x')`
  * ...her kan vi bruke `file` som kobling til den åpne filen...
  * ...vi trenger ikke indentering...
  * `file.close()`

````python
with open('telefon.txt') as tlf:
    print(tlf.closed)
    print(tlf.read(1))
    print(tlf.readline())
    print(tlf.read(1))
    print(tlf.read(0))
    print(tlf.readlines())
print(tlf.closed)
````

````python
# samme som:
tlf = open('telefon.txt')
print(tlf.closed)
print(tlf.read(1))
print(tlf.readline())
print(tlf.read(1))
print(tlf.read(0))
print(tlf.readlines())
tlf.close()
print(tlf.closed)
````

# Lesing med `for`-løkke
* En `fil` er en sekvens av linjer
  * `for linje in fil: ... `
  * går gjennom filen linje for linje

````python
tlf = open('telefon.txt')
for linje in tlf:
    print(linje.strip())
tlf.close()
````

````python
# litt mer kompakt:
for linje in open('telefon.txt'):
    print(linje.strip())
# ...men vi lukker jo ikke filen...
````

````python
# enkelt og mest elegant:
with open('telefon.txt') as fil:
    for linje in fil:
        print(linje.strip())
````

# Skriveoperasjoner
* `open(filnavn, 'w')`
  * oppretter en fil med oppgitt navn
  * hvis filen allerede finnes, blir den umiddelbart tømt
  * åpner for skriving til filen
* `open(filnavn, 'a')`
  * åpner for å legge til på slutten av eksisterende fil
  * _ingen overskriving_
  * hvis filen ikke finnes blir den opprettet
* `fil.write(tekststreng)`
  * utvider filen med tegnstrengen
  * husk: å ta med `\n` når du vil avslutte linjen
  * returnerer antall tegn som ble skrevet (men det har vi sjelden bruk for)
* Husk å bruke `fil.close()` eller `with open(filnavn, mode='x') as fil:`
  * for å være sikker på at filen lagres permanent på disken

# Unicode og UTF-8
* Hvordan kan vi representere skrifttegn som tall?
* __American Standard Code for Information Interchange__ (__ASCII__) (fra 1963)
  * 7 biter, gir opp til 128 tegn

<img src="F08/better_ascii_table.jpg" width=500/>

* __ISO-8859__ (__ISO-LATIN__)
  * 8 biter = 1 byte, gir opp til 256 tegn
  * _iso-8859-1_ (_iso-latin-1_): variant som dekker nord-europeiske alfabeter
* __Unicode__-standarden for skrifttegn (siden 1991)
  * brukt på veven og etter hvert overalt
  * ambisjon: å dekke alle nåværende og historiske språk
  * basert på _UCS_ og _UTF_
* __Universal Coded Character Set (UCS)__
  * hvilke tall (_kodepunkter_, _"code points"_) står for hvilke tegn?
  * opp til 1.1M tegn, &gt;144 000 i bruk så langt, &gt;150 skriftsystemer
  * _de første 128 "kodepunktene" er de samme som i ASCII!_
* __Unicode Transformation Format (UTF)__
  * hvordan skal vi overføre og lagre tallene (kodepunktene)?
  * flere varianter, som hver bruker ulike antall byte-r
  * __UTF-8__
    * varierende fra 1 til 4 byter
    * en sømløs utvidelse av ASCII
    * &gt;95% av alle vevsider
  * ...dessuten _UTF-16_ og _UTF-32_

`open(filnavn, encoding='kodesystem')`, f eks 
  * `open(filnavn, encoding='utf-8')`  # UTF-8 er standardverdien
  * `open(filnavn, encoding='iso-8859-1')`

# Eksempel: fra små til store bokstaver
* Lag et program som leser inn en fil og lager en ny fil med samme innhold skrevet med store bokstaver
* Tilleggsoppgave:
  * godta andre kodinger enn UTF-8

````python
def store_bokstaver(filnavn):
    innfil = open(filnavn)
    utfil = open('store-bokstaver-'+filnavn, 'w')
    for linje in innfil:
        utfil.write(linje.upper())
    innfil.close()
    utfil.close()
    
# test
fn = 'tre-små-kinesere-i-UTF-8.txt'
store_bokstaver(fn)
````

````python
!type store-bokstaver-tre-små-kinesere-i-UTF-8.txt
````

````python
# mer elegant versjon med 'with'
def store_bokstaver(filnavn):
    with (open(filnavn) as innfil, 
          open('store-bokstaver-'+filnavn, mode='w') as utfil):
        for linje in innfil:
            utfil.write(linje.upper())
    
# test
fn = 'tre-små-kinesere-i-UTF-8.txt'
store_bokstaver(fn)
````

````python
# versjon som leser og skriver ulike encoding-er
def store_bokstaver(filnavn, fra='utf-8', til='utf-8'):
    with (open(filnavn, encoding=fra) as innfil, 
          open('store-bokstaver-'+filnavn, mode='w', encoding=til) as utfil):
        for linje in innfil:
            utfil.write(linje.upper())
    
# test
fn = 'tre-små-kinesere-i-ISO-8859-1.txt'
store_bokstaver(fn, fra='iso-8859-1', til='utf-8')
````

# Mer objektorientert versjon

````python
class Bankkonto:
    kontonummer = None
    innehaver = None
    saldo = None

    def hent_fra_fil(self, fil):
        linje = fil.readline()
        if linje=='':
            raise Exception('Ikke flere konti')
        self.kontonummer = int(linje.strip())
        self.innehaver = fil.readline().strip()
        self.saldo = int(fil.readline().strip())

    def skriv_ut(self):
        print(self.innehaver, self.kontonummer, self.saldo)

    def legg_til_fil(self, fil):
        fil.write(f'{self.innehaver}\n{self.kontonummer}\n{self.saldo}\n')
````

````python
def hent_alle_fra_fil(filnavn):
    liste = []
    fil = open(filnavn)
    while True:
        ny_konto = Bankkonto()
        try:
            ny_konto.hent_fra_fil(fil)
        except:
            break
        liste.append(ny_konto)
    return liste
````

````python
def legg_alle_til_fil(kontoliste, filnavn):
    with open(filnavn, 'a') as fil:
        for konto in kontoliste:
            konto.legg_til_fil(fil)
````

````python
# test
filnavn = 'kontoer.txt'
kontoliste = hent_alle_fra_fil(filnavn)

ny_konto = Bankkonto()
ny_konto.kontonummer = 87654
ny_konto.innehaver = 'Andreas'
ny_konto.saldo = 9000
kontoliste.append(ny_konto)

legg_alle_til_fil(kontoliste, 'nye-'+filnavn)
````

````python
for k in kontoliste:
    k.skriv_ut()
````
